```python
# Cell 3: Run predictions and print results
print("=== CHURN PREDICTION TEST RESULTS ===\n")

for idx, customer in enumerate(sample_customers, 1):
    result = predict_churn_risk(customer)
    print(f"Customer #{idx}:")
    print(f"  • Risk Level       : {result['risk_level']}")
    print(f"  • Churn Probability: {result['churn_probability']:.2%}")
    print("  • Model Risk Factors:")
    for factor in result["top_risk_factors"]:
        print(f"      - {factor}")
    print("-" * 50)
```

```python
# Cell 4: View results in a DataFrame summary
import pandas as pd

summary_data = []
for idx, customer in enumerate(sample_customers, 1):
    res = predict_churn_risk(customer)
    summary_data.append(
        {
            "Customer": f"Customer #{idx}",
            "Risk Level": res["risk_level"],
            "Churn Probability": f"{res['churn_probability']:.2%}",
            "Top Risk Factor": res["top_risk_factors"][0]
            if res["top_risk_factors"]
            else "None",
        }
    )

pd.DataFrame(summary_data)
```
